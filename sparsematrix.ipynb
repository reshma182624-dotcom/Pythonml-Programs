```python
import numpy as np
from scipy.sparse import csr_matrix
matrix = np.array([[1, 0, 0],
                   [0, 0, 2],
                   [0, 3, 0]])
sparse_matrix = csr_matrix(matrix)
print("Given Matrix:")
print(matrix)
print("Sparse Matrix:")
print(sparse_matrix)
```

```python
import numpy as np
from scipy.sparse import csr_matrix
matrix = np.array([[1, 0],
                   [0, 0],
                   [0, 3]])
sparse_matrix = csr_matrix(matrix)
print("Matrix:")
print(matrix)
print("Sparse Matrix:")
print(sparse_matrix)
```

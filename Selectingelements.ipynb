```python
import numpy as np
vector = np.array([10,20,30,40,50,60])
matrix = np.array([[1, 2,3],
                   [3, 4,5]])
print("Vector:")
print(vector)
print("\nMatrix:")
print(matrix)
print("\nFirst element of vector is :", vector[0])
print("Second element of vector is :", vector[1])
print("Last element of vectoris :", vector[-1])
print("\nFirst row of the given matrix is :", matrix[0])
print("Second row of the given matrix is :", matrix[1])
print("First column:", matrix[:, 0])
print("Second column:", matrix[:, 1])
print("Element at first row and second column:", matrix[0, 1])
```
